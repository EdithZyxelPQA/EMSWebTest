```python
import tkinter as tk
import videocap_module as videocap

var = 0
video_file = "video_test.mp4"             ### only left window
file_name = video_file
t1 = videocap.VideoCapThread(video_file)

def vid():     
    print(f'clicked vi')
    t1.start()

def vid_stop():
    print(f'clicked vi_stop')
    t1.stop_record()


window = tk.Tk()
a1 = tk.Label(window,text=f"file_name:{file_name}")
a1.pack()
# e1 = tk.Entry(window,textvariable=file_name)
# e1.pack() 
b1 = tk.Button(window,text='vi',command=vid)
b1.pack()
b2 = tk.Button(window,text='vi_stop',command=vid_stop)
b2.pack()





window.mainloop()

```

```python
import tkinter as tk



def c1():     
    print(f'clicked vi')
    label.config(text='com1 ed')

def c2():
    print(f'clicked vi_stop')
    label.config(text='com2 ed')

def toggle_button(btn):
    if btn["relief"] == "raised":
        btn["relief"] = "sunken"
        btn["text"] = "沉"
    else:
        btn["relief"] = "raised"
        btn["text"] = "升"
        

window = tk.Tk()
label = tk.Label(window,text=f"click")
label.pack(pady=10)


# e1 = tk.Entry(window,textvariable=file_name)
# e1.pack() 
b1 = tk.Button(window,text='b1',command=lambda:(c1(),toggle_button(b1)),relief="raised")
b1.pack()
b2 = tk.Button(window,text='b2',command=c2)
b2.pack(pady=10)





window.mainloop()
```
